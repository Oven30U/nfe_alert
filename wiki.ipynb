```python
# Ejecutar bloques de tareas y almacenar en un dataframe los resultados de ambos bloques

# Crear un DataFrame vacío para almacenar los resultados
df = pd.DataFrame(columns=["Nombre", "Notificacion", "Screenshot", "Error"])

# Crear el primer bloque de tareas
tareas_bloque1 = [
    asyncio.create_task(instancia.procesar_jurisdiccion())
    for instancia in [nacional, agip, arba]
]

# Ejecutar el primer bloque de tareas y añadir los resultados al DataFrame
resultados_bloque1 = await asyncio.gather(*tareas_bloque1)
resultados_bloque1 = [list(res) for res in resultados_bloque1]
df_bloque1 = pd.DataFrame(resultados_bloque1, columns=df.columns)
df = df.append(df_bloque1, ignore_index=True)

# Crear el segundo bloque de tareas
tareas_bloque2 = [
    asyncio.create_task(instancia.procesar_jurisdiccion())
    for instancia in [mendoza, cordoba, neuquen, rio_negro]
]

# Ejecutar el segundo bloque de tareas y añadir los resultados al DataFrame
resultados_bloque2 = await asyncio.gather(*tareas_bloque2)
resultados_bloque2 = [list(res) for res in resultados_bloque2]
df_bloque2 = pd.DataFrame(resultados_bloque2, columns=df.columns)
df = df.append(df_bloque2, ignore_index=True)

print(df)
```

```python
# Using XPath
await page.fill('xpath=//input[@name="username"]', 'my_username')  # Fill an input with name "username"
await page.click('xpath=//button[contains(text(), "Submit")]')  # Click on a button with text "Submit"

# ID
element = await page.$x('//*[@id="elementID"]')
element = driver.find_element_by_xpath('//*[@id="elementID"]')

# Class
element = await page.$x('//*[@class="className"]')
element = driver.find_element_by_xpath('//*[@class="className"]')

# Etiqueta
element = await page.$x('//tagName')
element = driver.find_element_by_xpath('//tagName')

# Texto
element = await page.$x('//*[text()="elementText"]')
element = driver.find_element_by_xpath('//*[text()="elementText"]')

# Atributo
element = await page.$x('//*[@attribute="value"]')
element = driver.find_element_by_xpath('//*[@attribute="value"]')

# Primer elemento de un tipo en especifico - es importante colocarlo entre parentesis para indicar el inndice
element = await page.$x('(//tagName)[1]')
element = driver.find_element_by_xpath('(//tagName)[1]')

# Seleccionar por texto especifico
element = await page.$x('//*[contains(text(), "partOfText")]')
element = driver.find_element_by_xpath('//*[contains(text(), "partOfText")]')

# Por multiples condiciones
element = await page.$x('//tagName[@attribute="value"][text()="elementText"]')
element = driver.find_element_by_xpath('//tagName[@attribute="value"][text()="elementText"]')
```

# PyTest

```python
class BaseTest:
    """Test para el manejo coherente de todos los retornos"""

    @pytest.mark.asyncio
    async def run_base_test(
        self,
        Jurisdiccion,
        client,
        cuit,
        clave_fiscal,
        fecha_desde,
        fecha_hasta,
        cuit_cliente_input,
    ):
        async with async_playwright() as playwright:
            jurisdiccion = await Jurisdiccion.create(
                playwright,
                client,
                cuit,
                clave_fiscal,
                fecha_desde,
                fecha_hasta,
                cuit_cliente_input,
            )
            await jurisdiccion.procesar_jurisdiccion()

            assert jurisdiccion.page is not None
            assert jurisdiccion.hay_notificacion is not None
            assert jurisdiccion.hay_screenshot is not None
            if jurisdiccion.hay_notificacion not in [
                "Hay notificaciones",
                "No hay notificaciones",
            ] or jurisdiccion.hay_screenshot not in [
                "Se realizó screenshot",
                "No se realizó screenshot",
            ]:
                assert jurisdiccion.error is None
            assert jurisdiccion.nombre is not None
            if jurisdiccion.error is not None:
                assert (
                    jurisdiccion.hay_notificacion == "Error al tomar screenshot"
                    and jurisdiccion.hay_screenshot == "Error al tomar screenshot"
                )
```

```python
class TestNacional(BaseTest, ErrorTest):
    def setup(self):
        self.Jurisdiccion = Nacional
        self.client = "EDGE ARGENTINA S.R.L"
        self.cuit = "20386165476"
        self.clave_fiscal = "Gabriel1994"
        self.fecha_desde = "01/05/2024"
        self.fecha_hasta = "30/05/2024"
        self.cuit_cliente_input = "30714604356"

    @pytest.mark.asyncio
    async def test_nacional(self):
        await self.run_base_test(
            self.Jurisdiccion,
            self.client,
            self.cuit,
            self.clave_fiscal,
            self.fecha_desde,
            self.fecha_hasta,
            self.cuit_cliente_input,
        )

    @pytest.mark.asyncio
    async def test_nacional_error(self):
        await self.run_error_test(
            self.Jurisdiccion,
            self.client,
            self.cuit,
            self.clave_fiscal,
            self.fecha_desde,
            self.fecha_hasta,
            self.cuit_cliente_input,
        )
```
