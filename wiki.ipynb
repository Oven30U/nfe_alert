# XPATH

```python
# Using XPath
await page.fill('xpath=//input[@name="username"]', 'my_username')  # Fill an input with name "username"
await page.click('xpath=//button[contains(text(), "Submit")]')  # Click on a button with text "Submit"

# ID
element = await page.$x('//*[@id="elementID"]')
element = driver.find_element_by_xpath('//*[@id="elementID"]')

# Class
element = await page.$x('//*[@class="className"]')
element = driver.find_element_by_xpath('//*[@class="className"]')

# Etiqueta
element = await page.$x('//tagName')
element = driver.find_element_by_xpath('//tagName')

# Texto
element = await page.$x('//*[text()="elementText"]')
element = driver.find_element_by_xpath('//*[text()="elementText"]')

# Atributo
element = await page.$x('//*[@attribute="value"]')
element = driver.find_element_by_xpath('//*[@attribute="value"]')

# Primer elemento de un tipo en especifico - es importante colocarlo entre parentesis para indicar el inndice
element = await page.$x('(//tagName)[1]')
element = driver.find_element_by_xpath('(//tagName)[1]')

# Seleccionar por texto especifico
element = await page.$x('//*[contains(text(), "partOfText")]')
element = driver.find_element_by_xpath('//*[contains(text(), "partOfText")]')

# Por multiples condiciones
element = await page.$x('//tagName[@attribute="value"][text()="elementText"]')
element = driver.find_element_by_xpath('//tagName[@attribute="value"][text()="elementText"]')
```
