```python
import os
import sys
import pandas as pd

parent_dir = os.path.abspath(os.path.join(os.getcwd(), ".."))
sys.path.append(parent_dir)

from config import (
    PATH_ESTRUCTURA_ROBOT,
    NOMBRE_ARCHIVO_CLIENTE,
    SHEET_ARCHIVO_CLIENTE,
    TABLA_CLIENTE,
    TABLA_JURISDICCION,
)
```

```python
import os
import pandas as pd

def cargar_excels():
    dataframes = []
    
    for root, dirs, files in os.walk(PATH_ESTRUCTURA_ROBOT):
        for file in files:
            if file == NOMBRE_ARCHIVO_CLIENTE:
                file_path = os.path.join(root, file)
                
                # Leer el archivo de Excel utilizando pandas
                xls = pd.ExcelFile(file_path)
                
                # Verificar si la hoja existe
                if SHEET_ARCHIVO_CLIENTE in xls.sheet_names:
                    # Especificar manualmente el rango de filas y columnas para las tablas
                    skiprows_jurisdiccion = 4  # Ajusta esto según el número de filas a saltar para la tabla de jurisdicción
                    nrows_jurisdiccion = 27  # Ajusta esto según el número de filas en la tabla de jurisdicción
                    usecols_jurisdiccion = 'A:D'  # Ajusta esto según el rango de columnas de la tabla de jurisdicción
                    
                    skiprows_cliente = 0  # Ajusta esto según el número de filas a saltar para la tabla de cliente
                    nrows_cliente = 2  # Ajusta esto según el número de filas en la tabla de cliente
                    usecols_cliente = 'A:G'  # Ajusta esto según el rango de columnas de la tabla de cliente
                    
                    # Leer las tablas utilizando pandas
                    df_tabla_jurisdiccion = pd.read_excel(file_path, sheet_name=SHEET_ARCHIVO_CLIENTE, skiprows=skiprows_jurisdiccion, nrows=nrows_jurisdiccion, usecols=usecols_jurisdiccion)
                    df_tabla_cliente = pd.read_excel(file_path, sheet_name=SHEET_ARCHIVO_CLIENTE, skiprows=skiprows_cliente, nrows=nrows_cliente, usecols=usecols_cliente)
                    
                    # Repetir los valores de df_tabla_cliente para que coincidan con el número de filas de df_tabla_jurisdiccion
                    repeated_cliente_values = df_tabla_cliente.loc[df_tabla_cliente.index.repeat(len(df_tabla_jurisdiccion))].reset_index(drop=True)
                    
                    # Añadir las columnas de df_tabla_cliente a df_tabla_jurisdiccion
                    for col in df_tabla_cliente.columns:
                        df_tabla_jurisdiccion[col] = repeated_cliente_values[col]
                    
                    dataframes.append(df_tabla_jurisdiccion)

    if dataframes:
        df_final = pd.concat(dataframes, ignore_index=True)
        return df_final
    else:
        return pd.DataFrame()
```

```python
df_input = cargar_excels()
```
